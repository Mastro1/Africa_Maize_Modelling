```python
import pandas as pd

predictions = pd.read_csv(r"Model\africa_results\all_africa_maize_yield_predictions.csv")
countries = predictions["country"].unique().tolist()
print(countries)
```

```python
from pathlib import Path

folder = Path(r"Model_physical\Input")

csv_files = [p for p in folder.rglob("*.csv") if "new_ERA5" in p.name]

countries_downloaded = []

for p in csv_files:
    # Split once on the marker we know exists
    parts = p.name.split("_new_ERA5", 1)   # 1 = split only at first occurrence
    if len(parts) == 2:
        country_part = parts[0]            # "Angola_admin2" or "Kenya_admin1"
        # Remove the last _adminX part
        country = country_part.rsplit("_admin", 1)[0]
        countries_downloaded.append(country)

print("Missing countries:")
for c in countries:
    if c.replace(" ","_") not in countries_downloaded:
        print(c)

```
